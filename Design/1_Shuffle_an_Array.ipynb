## Solution 1: Using shuffled index list

```python
import random
import copy

class Solution(object):

    def __init__(self, nums):
        """
        :type nums: List[int]
        """
        self.original = nums
        self.nums = list(nums)

    def reset(self):
        """
        :rtype: List[int]
        """
        return self.original

    def shuffle(self):
        """
        :rtype: List[int]
        """
        index_li = range(len(self.nums))
        shuffled_index_li = random.sample(index_li, len(index_li))

        return [self.nums[i] for i in shuffled_index_li]

nums = [1,2,3,4,5,6]
obj = Solution(nums)
param_1 = obj.reset()
param_2 = obj.shuffle()

print(param_1)
print(param_2)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$

## Solution 2: Fisher-Yates shuffle

```python
import random
import copy

class Solution(object):

    def __init__(self, nums):
        """
        :type nums: List[int]
        """
        self.original = nums
        self.nums = list(nums)

    def reset(self):
        """
        :rtype: List[int]
        """
        return self.original

    def shuffle(self):
        """
        :rtype: List[int]
        """
        random.shuffle(self.nums)
        return self.nums

nums = [1,2,3,4,5,6]
obj = Solution(nums)
param_1 = obj.reset()
param_2 = obj.shuffle()

print(param_1)
print(param_2)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$
