## Solution 1: Using stack

```python
class Solution(object):
    def isValid(self, s):

        history = []
        open = ['(', '{', '[']
        close = [')', '}', ']']
        par_pair = {'(': ')',
               '{': '}', 
               '[': ']'}

        if s[-1] in open:
            return False

        for char in s:
            if char in open:
                history.append(char)
            elif char in close:
                if history: # There is no pair
                    return False
                if par_pair[history.pop()] != char: # pair is not right
                    return False

        if len(history) > 0: # open remain
            return False

        return True

sol = Solution()
s = '([]'
res = sol.isValid(s)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$

## Solution 2: Upgrade version by AI

```python
class Solution(object):
    def isValid(self, s):
        stack = []
        mapping = {')':'(', '}':'{', ']':'['}

        for char in s:
            if char in mapping:
                top_element = stack.pop() if stack else '#'
                if mapping[char] != top_element:
                    return False
            else:
                stack.append(char)
        return not stack

sol = Solution()
s = '([])'
res = sol.isValid(s)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$
