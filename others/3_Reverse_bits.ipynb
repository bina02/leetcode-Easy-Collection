## Solution 1: Using format and int function (decimal -> binary(reversed) -> decimal)

```python
class Solution(object):
    def reverseBits(self, n):
        binary = format(n, '032b')[::-1]
        return int(binary, 2)

sol = Solution()
n =  43261596
res = sol.reverseBits(n)
print(res)
```

Time Complexity: $O(\text{log}_2N)$ (O(1) in a 32-bit context)

Space Complexity: $O(1)$

```python
class Solution(object):
    def reverseBits(self, n):
        reversed = 0
        for _ in range(32):
            reversed = (reversed << 1) | (n & 1)
            n >>= 1
        return reversed

sol = Solution()
n =  43261596
res = sol.reverseBits(n)
print(res)
```

Time Complexity: $O(\text{log}_2N)$ (O(1) in a 32-bit context)

Space Complexity: $O(1)$ (O(1) in a 32-bit context)
