## Solution 1: Using XOR 

```python
class Solution(object):
    def count_one_bit(self, x):
        count = 0
        while x // 2 > 0:
            r = x % 2
            x = x // 2

            if r == 1:
                count += 1
        if x == 1:
            count += 1
        return count

    def hammingDistance(self, x, y):
        xor = x ^ y
        return self.count_one_bit(xor)

x = 1
y = 4
sol = Solution()
res = sol.hammingDistance(x, y)
print(res)
```

Time Complexity: $O(\text{log}_2N)$

Space Complexity: $O(1)$
