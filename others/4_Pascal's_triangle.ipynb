# Solution 1: Using rightsum & leftsum

```python
class Solution(object):
    def next_step(self, x):
        length = len(x)
        next_li = [0] * (length+1)

        for i in range(length):
             next_li[i] += x[i]
             next_li[i+1] += x[i]             
        return next_li

    def generate(self, numRows):
        triangle = [[1]]
            
        for i in range(numRows-1):
                next = self.next_step(triangle[i])
                triangle.append(next)

        return triangle

sol = Solution()
res = sol.generate(5)
print(res)
        
```

Time Complexity: $O(N^2)$

Space Complexity: $O(N^2)$
