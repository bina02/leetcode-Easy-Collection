## Solution 1: Bit Counting via Division

```python
class Solution(object):
    def hammingWeight(self, n):
        count = 0
        while n > 1:
            #print(n)
            r = n % 2
            n = n // 2
            if r == 1:
                count += 1

        if n == 0:
            return count
        elif n == 1:
            return count + 1

sol = Solution()
res = sol.hammingWeight(2)
print(res)
```

Time Complexity: $O(\text{log}_2n)$ 

Space Complexity: $O(1)$

## Solution 2: Making binary list and counting

```python
class Solution(object):
    def hammingWeight(self, n):
        count = 0
        binary_str = ''
        while n > 1:
            #print(n)
            r = n % 2
            n = n // 2
            if r == 1:
                binary_str += '1'
            if r == 0:
                binary_str += '0'

        if n == 1:
            binary_str += '1'
        binary_str = binary_str[::-1]
        return binary_str.count('1')

sol = Solution()
res = sol.hammingWeight(11)
print(res)
```

Time Complexity: $O(\text{log}_2n)$ 

Space Complexity: $O(\text{log}_2n)$ 

## Solution 3: Kernighan's Algorithm

```python
class Solution(object):
    def hammingWeight(self, n):
        count = 0
        while n > 0:
            n = n & (n-1)
            count += 1

        return count

sol = Solution()
res = sol.hammingWeight(14)
print(res)
```

Time Complexity: $O(K)$ K is number of 1 in binary

Space Complexity: $O(1)$
