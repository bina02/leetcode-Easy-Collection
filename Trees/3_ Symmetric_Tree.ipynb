```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

leaf6 = TreeNode(3)
leaf5 = TreeNode(4)
leaf4 = TreeNode(4)
leaf3 = TreeNode(3)
leaf2 = TreeNode(2, leaf5, leaf6)
leaf1 = TreeNode(2, leaf3, leaf4)
root = TreeNode(1, leaf1, leaf2)
```

## Solution 1: Using queue

```python
from collections import deque

class Solution(object):
    def isSymmetric(self, root):
        node_queue = deque([root])
        val_li = []
        while node_queue:
            for _ in range(len(node_queue)):
                node = node_queue.popleft()

                if node.left:
                    node_queue.append(node.left)
                    val_li.append(node.left.val)
                else:
                    val_li.append(None)

                if node.right:
                    node_queue.append(node.right)
                    val_li.append(node.right.val)
                else:
                    val_li.append(None)
            if val_li != val_li[::-1]:
                return False
            val_li = []
        return True
                    
sol = Solution()
result = sol.isSymmetric(root)
print(result)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$

## Solution 2: Using Recursion

```python
from collections import deque

class Solution(object):
    def isSymmetric(self, root):
        left = root.left
        right = root.right

        return self.isSym(left, right)
        
    def isSym(self, left, right):
        if not left and not right:
            return True
        
        if not left or not right or left.val != right.val:
            return False
        
        return self.isSym(left.left, right.right) and self.isSym(left.right, right.left)
                    
sol = Solution()
result = sol.isSymmetric(root)
print(result)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(h)$ h is number of stack
