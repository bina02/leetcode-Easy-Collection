```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# example 1:
leaf1 = TreeNode(7)
leaf2 = TreeNode(15)
leaf3 = TreeNode(20, leaf2, leaf1)
leaf4 = TreeNode(9)
root = TreeNode(3, leaf4, leaf3)
```

## Solution 1: DFS

```python
class Solution(object):
    def maxDepth(self, root):
        if not root:
            return 0
        left = self.maxDepth(root.left)
        right = self.maxDepth(root.right)

        return 1 + max(left, right)

sol = Solution()
result = sol.maxDepth(root)
print(result)
        
```

Time Complexity: $O(n)$ 

Space Complexity: $O(h)$ $h$ is number of stack

## Solution 2: BFS

```python
from collections import deque

class Solution(object):
    def maxDepth(self, root):
        if not root:
            return 0

        depth = 0
        node_queue = deque([root])
        while node_queue:
            for _ in range(len(node_queue)):
                current_node = node_queue.popleft()
                if current_node.left:
                    node_queue.append(current_node.left)
                if current_node.right:
                    node_queue.append(current_node.right)
            depth += 1
        return depth

sol = Solution()
result = sol.maxDepth(root)
print(result)
        
```

Time Complexity: $O(n)$ 

Space Complexity: $O(w)$ $w$ is number of queue
