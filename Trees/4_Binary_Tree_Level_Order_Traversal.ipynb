```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
leaf4 = TreeNode(7)
leaf3 = TreeNode(15)
leaf2 = TreeNode(20, leaf3, leaf4)
leaf1 = TreeNode(9)
root = TreeNode(3, leaf1, leaf2)
```

## Solution 1: Using queue

```python
from collections import deque

class Solution(object):
    def levelOrder(self, root):
        if not root:
            return []

        queue = deque([root])
        val_list = []
        while queue:
            val_li = []
            for _ in range(len(queue)):
                node = queue.popleft()
                val_li.append(node.val)
                if node.left:
                   queue.append(node.left)
                if node.right:
                    queue.append(node.right)
            val_list.append(val_li)
        
        return val_list

sol = Solution()
result = sol.levelOrder(root)
print(result)
        
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$
