### Solution1: list->int->str->list

```python
class Solution1(object):
    def plusOne(self, digits):
        num = 0
        result = []
        digits.reverse()

        # int to list
        for i in range(len(digits)):
            num += (digits[i] * 10**i)

        # int + 1
        num += 1

        # int to list
        for s in str(num):
            result.append(int(s))
        return result

solution = Solution1()

digits = [9]
result = solution.plusOne(digits)
print(result)
```

Time Complexity is $O(n^2)$

### Solution2: carry in list

```python
class Solution2(object):
    def plusOne(self, digits):
        length = len(digits)

        for i in range(length -1, -1, -1):
            # digits[i] is not 9
            if digits[i] < 9:
                digits[i] += 1
                return digits
            
            # digits[i] is 9
            digits[i] = 0
        # catch digits = [9,9,9...]
        return [1] + digits
    
solution = Solution2()
digits = [9, 9]
result = solution.plusOne(digits)
print(result)
```

## Comparison: Solution1 vs. Solution2

```python
import time
import numpy as np

digits = np.arange(1,4000).tolist()

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    result = sol.plusOne(digits)
    end = time.time()
    print(np.round(end - start, 4))
```

Solution1 has a problem: the method using str() fails when the length of digits exceeds 4300.
