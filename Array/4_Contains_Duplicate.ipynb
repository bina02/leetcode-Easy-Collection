### Solution 1. For & in

```python
class Solution1(object):
    def containsDuplicate(self, nums):
        for i, num in enumerate(nums):
            if num in nums[i+1:]:
                return True
        return False
    
solution = Solution1()
nums = [1,2,3,4,1]
result = solution.containsDuplicate(nums)
print(result)
```

Time Complexity = $O(n^2)$

### Solution 2. Hashmap

```python
class Solution2(object):
    def containsDuplicate(self, nums):
        map = set()

        for num in nums:
            if num in map:
                return True
            map.add(num)
        return False
        # return len(nums) != len(set(nums)) is faster too!

solution2 = Solution2()
nums = [1,2,3,4,1]
result = solution2.containsDuplicate(nums)
print(result)
```

Time Complexity = $O(n)$

## Comparison: Solution1 vs. Solution2

```python
import time
import numpy as np

solution1 = Solution1()
solution2 = Solution2()

nums = np.arange(100000).tolist()

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    result = sol.containsDuplicate(nums)
    end = time.time()
    print(np.round(end - start, 4))

```
