## Solution 1: sliding window

```python
class Solution1(object):
    def strStr(self, haystack, needle):
        left = 0
        right = len(needle)
        for i in range(len(haystack) - len(needle) + 1):
            if haystack[left:right] == needle:
                return left
            left += 1
            right += 1
        
        return -1

solution = Solution1()
haystack = 'sadbutsad'
needle = 'sad'
result = solution.strStr(haystack, needle)
print(result)
```

Time complexity : $O(n \times m)$

## Solution 2: Rabin-Karp

```python
class Solution2(object):
    def strStr(self, haystack, needle):
        hay_len = len(haystack)
        need_len = len(needle)

        if hay_len < need_len:
            return -1

        target_hash = sum(ord(c) for c in needle)
        current_hash = sum(ord(haystack[i]) for i in range(need_len))
        
        for i in range(hay_len - need_len + 1):
            if current_hash == target_hash:
                if haystack[i:i+need_len] == needle:
                    return i
                
            if i < hay_len - need_len:
                current_hash = current_hash - ord(haystack[i]) + ord(haystack[i+need_len])

        return -1

solution = Solution2()
haystack = 'sdbutsad'
needle = 'sad'
result = solution.strStr(haystack, needle)
print(result)
```

Time complexity: Best $O(n)$, Worst $O(n \times m)$

```python
import time
import numpy as np

# create data
N = 1000000
M = 50000
haystack = 'a' * N + 'b' * M
needle = 'a' * 10 + 'b' * M

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    sol.strStr(haystack, needle)
    end = time.time()
    print(f"{sol.__class__.__name__}: {round(end - start, 4)}s")
```
