## Solution 1: Using slicing

```python
class Solution1(object):
    def firstUniqChar(self, s):
        for i in range(len(s)):
            if s[i] not in s[:i] + s[i+1:]:
                return i
        return -1

solution = Solution1()
s = 'loveleetcode'
result = solution.firstUniqChar(s)
print(result)
```

Time Complexity: $O(n^2)$  
Space Complexity: $O(n)$

## Solution 2: Using hashmap

```python
class Solution2(object):
    def firstUniqChar(self, s):
        map = {}
        for char in s:
            map[char] = map.get(char, 0) + 1
        
        for i in range(len(s)):
            if map[s[i]] == 1:
                return i
        return -1

solution = Solution2()
s = 'loveleetcode'
result = solution.firstUniqChar(s)
print(result)
```

Time Complexity: $O(n)$  
Space Complexity: $O(n)$

## Solution1 vs. Solution2

```python
import time
import numpy as np

# create data
s = ''
for _ in range(500000):
    s += 'a'
s += 'b'

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    sol.firstUniqChar(s)
    end = time.time()
    print(f"{sol.__class__.__name__}: {round(end - start, 4)}s")
```
