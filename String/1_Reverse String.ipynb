## Solution 1: Two pointer

```python
class Solution1(object):
    def reverseString(self, s):
        left = 0
        right = len(s)-1

        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1
        return s
    
solution = Solution1()
s = ["h","e","l","l","o"]
result = solution.reverseString(s)
print(result)
```

Time Complexity: $O(n)$  
Space Complexity: $O(1)$

## Solution 2: Slicing

```python
class Solution2(object):
    def reverseString(self, s):
        return s[::-1]
    
solution = Solution2()
s = ["h","e","l","l","o"]
result = solution.reverseString(s)
print(result)
```

Time Complexity: $O(n)$  
Space Complexity: $O(n))$

## Solution 1 vs. Solution 2

```python
import time
import numpy as np

# create data
s = []
for _ in range(10000000):
    s.append('a')

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    sol.reverseString(s)
    end = time.time()
    print(np.round(end - start, 4))


```

Solution2 is faster than Solution1, because slicing function is implemented by C
