```python
class Solution(object):
    def myAtoi(self, s):
        #1. Whitespace
        s = s.lstrip()
        if not s:
            return 0

        #2. Signedness
        sign = 1
        if s[0] == '-':
            sign = -1
            s = s[1::]
        elif s[0] == '+':
            s = s[1::]

        #3. Conversion
        result = 0
        for i in range(len(s)):
            if s[i].isdigit():
                result = result * 10 + int(s[i])
            else:
                break

        result = result * sign
    
        #4. Rounding
        min, max = -2**31, 2**31 - 1
        if result < min:
            result = min
        if result > max:
            result = max
        
        return result

s = 'words and 987'
solution = Solution()
result = solution.myAtoi(s)
print(result)
```

Time Complexity: $O(n)$
