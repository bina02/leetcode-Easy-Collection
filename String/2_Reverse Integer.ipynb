## Solution 1: int -> list -> int

```python
class Solution1(object):
    def reverse(self, x):
        result = 0
        sign = 1
        li = list(str(x))

        # if x < 0, remove '-' and muptiplies after
        if x < 0:
            sign = sign * (-1)
            li.pop(0)

        li = li[::-1]
        for i in range(len(li)):
            result += int(li[i]) * 10**(len(li)-1-i)

        result = result * sign
        
        # according to constrait, function should return 0 if -2^31 -1 < x < 2^31 -1
        if result < -2**31 or result > (2**31 - 1):
            return 0

        return result

solution = Solution1()
x = 213
result = solution.reverse(x)
print(result)
```

Time Complexity: $O(n)$  
Space Complexity: $O(n)$

## Solution 2: Using Quotient & Remainder

```python
class Solution2(object):
    def reverse(self, x):
        sign = 1 if x >= 0 else -1
        result = 0
        x = abs(x)

        while x != 0:
            r = x % 10
            x = x // 10

            result = result * 10 + r
        
        if result < -2**31 or result > (2**31 - 1):
            return 0
        
        return result * sign

solution = Solution2()
x = 213
result = solution.reverse(x)
print(result)
```

Time Complexity: $O(n)$  
Space Complexity: $O(1)$

## Solution1 vs. Solution2

```python
import time
import numpy as np

# create data
x = 2**10000

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    sol.reverse(x)
    end = time.time()
    print(f"{sol.__class__.__name__}: {round(end - start, 4)}s")
```
