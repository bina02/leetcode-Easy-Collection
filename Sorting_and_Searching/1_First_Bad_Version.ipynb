## Solution 1: Finding with for loop

```python
# The isBadVersion API is already defined for you.
# @param version, an integer
# @return a bool
# def isBadVersion(version):

class Solution(object):
    def firstBadVersion(self, n):
        for num in range(1, n+1):
            if isBadVersion(num):
                return num
```

Time Complexity : $O(n)$

Space Complexity : $O(1)$

## Solution 2: Two pointer

```python
# The isBadVersion API is already defined for you.
# @param version, an integer
# @return a bool
# def isBadVersion(version):

class Solution(object):
    def firstBadVersion(self, n):
        min = 1
        max = n
        if isBadVersion(min):
            return min
        
        while not (isBadVersion(n) == True and isBadVersion(n-1) == False):
            n = (max + min) // 2
            if isBadVersion(n):
                max = n
            else:
                min = n
        return n
```

Time Complexity : $O(\text{log}\ n)$

Space Complexity : $O(1)$

## Solution 3: Two pointer(upgraded)

```python
# The isBadVersion API is already defined for you.
# @param version, an integer
# @return a bool
# def isBadVersion(version):

class Solution(object):
    def firstBadVersion(self, n):
        min = 1
        max = n

        while min < max:
            middle = (min + max) // 2
            if isBadVersion(middle):
                max = middle
            else:
                min = middle + 1
        return min
```

Time Complexity : $O(\text{log}\ n)$

Space Complexity : $O(1)$
