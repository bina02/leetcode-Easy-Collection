## Solution 1: Using two pointer

```python
class Solution(object):
    def merge(self, nums1, m, nums2, n):
        left = 0
        right = m
        nums1[m:] = nums2

        for _ in range(len(nums1)):
            for _ in range(n):
                if right > left and nums1[left] >= nums1[right]:
                    nums1[left], nums1[right] = nums1[right], nums1[left]
                right += 1
            left += 1
            right = m
        return nums1
```

```python
nums1 = [4,5,6,0,0,0]
m = 3
nums2 = [1,2,3]
n = 3
sol = Solution()
result = sol.merge(nums1, m, nums2, n)
print(result)
```

Time Complexity : $O(n \times m)$

Space Complexity : $O(1)$

## Solution 2: Advenced two pointer(Three pointer)

```python
class Solution(object):
    def merge(self, nums1, m, nums2, n):
        p1 = m-1
        p2 = n-1
        p = m+n-1

        while p1 >= 0 and p2 >= 0:
            if nums1[p1] > nums2[p2]:
                nums1[p] = nums1[p1]
                p1 -= 1

            else:
                nums1[p] = nums2[p2]
                p2 -= 1
            p -= 1

        while p2 >= 0:
            nums1[p] = nums2[p2]
            p -= 1
            p2 -= 1

        return nums1
```

```python
nums1 = [4,5,6,0,0,0]
m = 3
nums2 = [1,2,9]
n = 3
sol = Solution()
result = sol.merge(nums1, m, nums2, n)
print(result)
```

Time Complexity : $O(n + m)$

Space Complexity : $O(1)$
