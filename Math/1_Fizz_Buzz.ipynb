## Solution 1: deviding case with elif

```python
class Solution(object):
    def fizzBuzz(self, n):
        num_li = []
        for num in range(1, n+1):
            if num % 15 == 0:
                num_li.append('FizzBuzz')
            elif num % 3 == 0:
                num_li.append('Fizz')
            elif num % 5 == 0:
                num_li.append('Buzz')
            else:
                num_li.append(str(num))
        return num_li

sol = Solution()
res = sol.fizzBuzz(5)
print(res)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$
