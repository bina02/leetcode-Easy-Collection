## Solution 1: Using recursion

```python
class Solution(object):
    def transform(self, symbol):
        values = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
        return values.get(symbol)

    def recursion(self, s_list, i=0):
        if i == len(s_list)-1:
            return self.transform(s_list[i])
        
        current = self.transform(s_list[i])
        next = self.transform(s_list[i+1])

        if current >= next:
            return current + self.recursion(s_list, i+1)
        else:
            return -current + self.recursion(s_list, i+1)
          
    def romanToInt(self, s):
          return self.recursion(list(s))
        
sol = Solution()
s = "MCMXCIV"
res = sol.romanToInt(list(s))
print(res)

```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$

## Solution 2: For loop

```python
class Solution(object):
    def transform(self, symbol):
        values = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
        return values.get(symbol)

    def romanToInt(self, s):
        total = 0
        n = len(s)
        for i in range(n-1):
            current = self.transform(s[i])
            next = self.transform(s[i+1])
            if current >= next:
                total += current
            else:
                total -= current
        total += self.transform(s[-1])

        return total

sol = Solution()
s = "MCMXCIV"
res = sol.romanToInt(list(s))
print(res)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(1)$
