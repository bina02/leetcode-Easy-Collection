## Solution 1: Finding prime number with previous numbers

```python
class Solution(object):
    def isPrimeNumber(self, number):
        if number == 1:
            return False
        if number == 2:
            return True 

        for num in range(2, number):
            if number % num == 0:
                return False
        return True

    def countPrimes(self, n):
        count = 0
        for num in range(1, n):
            if self.isPrimeNumber(num):
                count += 1
        return count

sol1 = Solution()
n = 10
res = sol1.countPrimes(n)
print(res)
```

Time Complexity: $O(n^2)$ 

Space Complexity: $O(1)$

## Solution 2: Sieve of Eratosthenes

```python
class Solution(object):
    def countPrimes(self, n):
        if n <= 1:
            return 0
        
        li = [True] * n
        li[0] = li[1] = False

        for i in range(2, int(n**(0.5)+1)):
            if li[i]:
                for j in range(i*i, n, i):
                    li[j] = False

        return sum(li)

sol2 = Solution()
n = 10
res = sol2.countPrimes(n)
print(res)
```

Time Complexity: $O(n\text{loglog}n)$ 

Space Complexity: $O(n)$

## Compare Sol1 vs Sol2

```python
import time
def measureTime(sol, n):
    start = time.time()
    res = sol.countPrimes(n)
    end = time.time()
    print(f'result: {res}, time: {(end - start):.2f}s')

measureTime(sol1, 60000)
measureTime(sol2, 100000000)
```
