## Solution 1. Checking multiplies of 3 and Divide by 3

```python
class Solution(object):
    def isPowerOfThree(self, n):
        if n <= 0:
            return False
        
        while n % 3 == 0:
            n //= 3
        return n == 1

sol = Solution()
n = 45
res = sol.isPowerOfThree(n)
print(res)
```

Time Complexity: $O(\text{log}_3n)$ 

Space Complexity: $O(1)$
