## Solution1: Brute Force

```python
class Solution1(object):
    def intersect(self, nums1, nums2):
        result = []
        for num in nums1:
            if num in nums2:
                result.append(num)
                nums2.remove(num) 
        return result

solution = Solution1()
nums1 = [1,2,2,1]
nums2 = [2,2]
print(solution.intersect(nums1, nums2))
```

Time Complexity is $O(n^2)$

## Solution2: Using Hashmap

```python
class Solution2(object):
    def intersect(self, nums1, nums2):
        counts = {}
        result = []
        for num in nums1:
            counts[num] = counts.get(num, 0) + 1
        
        for num in nums2:
            if counts.get(num, 0) > 0:
                result.append(num)
                counts[num] -= 1

        return result

solution = Solution2()
nums1 = [4,9,5]
nums2 = [9,4,9,8,4]
print(solution.intersect(nums1, nums2))
```

Time Complexity is $O(n + m)$
- n is len of nums1
- m is len of nums2

## Comparison: Solution1 vs. Solution2

```python
import time
import numpy as np

nums1 = np.arange(200000).tolist()
nums2 = np.arange(200000).tolist()

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    result = sol.intersect(nums1, nums2)
    end = time.time()
    print(np.round(end - start, 4))
```
