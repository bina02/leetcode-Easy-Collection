### Solution 1: remove & append

```python
class Solution1(object):
    def moveZeroes(self, nums):
        num_zero = 0
        for _ in range(len(nums)):
            if 0 in nums:
                nums.remove(0)
                num_zero += 1
        
        for _ in range(num_zero):
            nums.append(0)

        return nums
            

solution = Solution1()
nums = [0,1,0,3,12]
result = solution.moveZeroes(nums)
print(result)
```

Time Complexity: $O(n^2)$

### Solution 2: Two pointer

```python
class Solution2(object):
    def moveZeroes(self, nums):
        j = 0
        for i in range(len(nums)):
            if nums[i] != 0:
                nums[j] = nums[i]
                j += 1
        
        for j in range(j, len(nums)):
            nums[j] = 0

        return nums
    
solution = Solution2()
nums = [0,1,0,3,12]
result = solution.moveZeroes(nums)
print(result)
```

Time Complexity: $O(n)$

## Comparison: Solution1 vs. Solution2

```python
import time
import numpy as np

nums = np.arange(0, 10000000).tolist()

sols = [Solution1(), Solution2()]
for sol in sols:
    start = time.time()
    result = sol.moveZeroes(nums)
    end = time.time()
    print(np.round(end - start, 4))
```
