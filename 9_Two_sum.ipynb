## Solution1: Brute force

```python
class Solution1(object):
    def twoSum(self, nums, target):
        for i in range(len(nums)):
            for j in range(i+1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]
    
nums = [2,7,11,15]
target = 9

solution = Solution1()
result = solution.twoSum(nums, target)
print(result)
```

Time Complexity: $O(n^2)$

## Solution2: using difference with in

```python
class Solution2(object):
    def twoSum(self, nums, target):
        for i in range(len(nums)):
            diff = target - nums[i]
            if diff in nums[i+1:]:
                index = nums[i+1:].index(diff) + (i+1)
                return [i, index]
        
    
nums = [2,7,11,15]
target = 9

solution = Solution2()
result = solution.twoSum(nums, target)
print(result)
```

Time complexity: $O(n^2)$

## Solution 3:

```python
class Solution3(object):
    def twoSum(self, nums, target):
        map = {}
        
        for i, n in enumerate(nums):
            diff = target - n
            if diff in map:
                return [map[diff], i]
            map[n] = i


    
nums = [2,7,11,15]
target = 9

solution = Solution3()
result = solution.twoSum(nums, target)
print(result)
```
