## Solution1 : Using binary tree

```python
from collections import deque

class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution(object):
    def make_tree(self, nums, index):
        if index >= len(nums):
            return None
        return TreeNode(nums[index], 
                        self.make_tree(nums, index+2),
                        self.make_tree(nums, index+3))

    def max_value(self, root):
        if not root:
            return 0
        
        if not root.left and not root.right:
            return root.val

        if root.left:
            left_max = self.max_value(root.left)
        else:
            left_max = float('-inf')

        if root.right:
            right_max = self.max_value(root.right)
        else:
            right_max = float('-inf')

        return root.val + max(left_max, right_max)
    
    def rob(self, nums):
        tree1 = self.make_tree(nums, 0)
        tree2 = self.make_tree(nums, 1)
        
        return max(self.max_value(tree1), self.max_value(tree2))
        

sol = Solution()
nums = [1,2,3,4,5,6,7]
res = sol.rob(nums)
print(res)
```

Time Complexity : $O(n)$

Space Complexity : $O(n)$

## Solution 2: Using Recurrence (Recursion)

```python
class Solution:
    def rob(self, nums):
        if len(nums) == 0:
            return 0
        if len(nums) == 1:
            return nums[0]
        if len(nums) == 2:
            return max(nums)

        return max(self.rob(nums[:-2]) + nums[-1], self.rob(nums[:-1]))

sol = Solution()
nums = [1,2,3,4,5]
res = sol.rob(nums)
print(res)
```

Time Complexity : $O(2^n)$

Space Complexity : $O(n)$

## Solution 3: Using Recurrence

```python
class Solution:
    def rob(self, nums):
        prev1 = 0
        prev2 = 0

        for num in nums:
            current = max(prev2 + num, prev1)
            prev2 = prev1
            prev1 = current

        return current

sol = Solution()
nums = [1,2,3,4,5]
res = sol.rob(nums)
print(res)
```

Time Complexity : $O(n)$

Space Complexity : $O(1)$
