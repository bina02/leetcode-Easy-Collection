## Solution 1: Permutations with repetition

```python
class Solution(object):
    def factorial(self, n):
        res = 1
        for num in range(1, n+1):
            res = res * num
        return res
    
    def rem_one_add_two(self, li, k):
        for _ in range(2 * k):
            li.pop()
        for _ in range(k):
            li.append(2)
        return li

    def create_one_list(self, k):
        li = []
        for _ in range(k):
            li.append(1)
        return li

    def climbStairs(self, n):
        count = 1

        iter = n // 2
        for k in range(1, iter+1):
            one_list = self.create_one_list(n)
            one_count = n
            two_count = 0
            li = self.rem_one_add_two(one_list, k)
            one_count = n - 2 * k
            two_count = k
            count += self.factorial(len(li)) / (self.factorial(one_count) * self.factorial(two_count))

        return int(count)

sol = Solution()
n = 10
result = sol.climbStairs(n)
print(result)
```

Time Complexity : $O(n)$

Space Complexity : $O(n)$

## Solution 2: Permutations with repetition(Upgraded)

```python
class Solution(object):
    def factorial(self, n):
        res = 1
        for num in range(1, n+1):
            res = res * num
        return res
    
    def climbStairs(self, n):
        count = 1

        iter = n // 2
        for k in range(1, iter + 1):
            one_count = n - 2*k
            two_count = k

            count = count + self.factorial(one_count + two_count) // (self.factorial(one_count) * self.factorial(two_count))

        return count

sol = Solution()
n = 10
result = sol.climbStairs(n)
print(result)
```

Time Complexity : $O(n)$

Space Complexity : $O(1)$

## Solution 3: Using Fibonacci sequence

```python
class Solution(object):
    def climbStairs(self, n):
        if n <= 2:
            return n
        
        prev1 = 1
        prev2 = 2

        for _ in range(3, n+1):
            current = prev1 + prev2
            prev1 = prev2
            prev2 = current

        return current

sol = Solution()
n = 10
result = sol.climbStairs(n)
print(result)
```

Time Complexity : $O(n)$

Space Complexity : $O(1)$
