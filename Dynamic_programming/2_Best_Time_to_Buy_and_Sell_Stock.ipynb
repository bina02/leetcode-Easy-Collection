## Solution 1: Finding max_value in sliced array

```python
class Solution(object):
    def maxProfit(self, prices):
        profit = 0
        for i in range(len(prices)-1):
            cur_price = prices[i]
            fut_prices = prices[i+1:]
            cur_profit = max(fut_prices) - cur_price
            if cur_profit < 0:
                cur_profit = 0
            
            profit = max(cur_profit, profit)

        return profit

sol = Solution()
prices = [7,1,5,3,6,4]
res = sol.maxProfit(prices)
print(res)
```

Time Complexity : $O(n^2)$

Space Complexity : $O(n)$

```python
class Solution(object):
    def maxProfit(self, prices):
        max_profit = 0
        min_price = prices[0]

        for i in range(1, len(prices)):
            max_profit = max(max_profit, prices[i] - min_price)
            min_price = min(prices[i], min_price)
        return max_profit

sol = Solution()
prices = [7,1,5,3,6,4]
res = sol.maxProfit(prices)
print(res)
```

Time Complexity : $O(n)$

Space Complexity : $O(1)$
