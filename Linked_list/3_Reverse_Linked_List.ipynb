```python
class Node:
    def __init__(self, x):
        self.val = x
        self.next = None

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)
head.next.next.next.next = Node(5)
```

## Solution1: Using list and reversed list

```python
class Solution(object):
    def reverseList(self, head):

        if not head:
            return None
        
        nodelist = []
        current_node = head
        while current_node:
            nodelist.append(current_node)
            current_node = current_node.next

        nodelist = nodelist[::-1]

        for i in range(len(nodelist) - 1):
            current_node = nodelist[i]
            current_node.next = nodelist[i+1]
            current_node = current_node.next

        nodelist[-1].next = None

        return nodelist[0]
```

```python
sol = Solution()
sol.reverseList(head)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$

## Solution2: Using pointer

```python
class Solution(object):
    def reverseList(self, head):
        if not head:
            return None
        
        prev = None
        current_node = head

        while current_node:
            next_node = current_node.next
            current_node.next = prev

            prev = current_node
            current_node = next_node

        return prev
```

```python
sol = Solution()
sol.reverseList(head)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(1)$
