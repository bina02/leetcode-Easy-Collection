```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

head1 = ListNode(1)
head1.next = ListNode(2)
head1.next.next = ListNode(2)
head1.next.next.next = ListNode(1)

head2 = ListNode(1)
head2.next = ListNode(2)
```

## Solution 1: Compare with reversed list

```python
class Solution(object):
    def isPalindrome(self, head):
        val_li = []
        
        while head:
            val_li.append(head.val)
            head = head.next

        return val_li == val_li[::-1]

sol = Solution()
sol.isPalindrome(head2)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(n)$

## Solution 2: Two pointer

```python
class Solution(object):
    def isPalindrome(self, head):
        slow, fast = head, head

        # find middle point
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # make reversed list
        prev = None
        current_node = slow

        while current_node:
            next_node = current_node.next
            current_node.next = prev
            prev = current_node
            current_node = next_node
        
        # compare
        current_node = head
        while prev:
            if current_node.val != prev.val:
                return False
            current_node = current_node.next
            prev = prev.next
        
        return True
    
sol = Solution()
sol.isPalindrome(head1)
```

Time Complexity: $O(n)$ 

Space Complexity: $O(1)$
