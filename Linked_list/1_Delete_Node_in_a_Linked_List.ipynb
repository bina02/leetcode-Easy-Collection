
## Solution 1:

```python
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

def printLinkedList(head):
    current = head
    result = []
    while current:
        result.append(str(current.val))
        current = current.next
    print(" -> ".join(result))

list_head = ListNode(head[0])
current_node = list_head

for i in range(1, len(head)):
    current_node.next = ListNode(head[i])
    current_node = current_node.next

printLinkedList(list_head)
```
