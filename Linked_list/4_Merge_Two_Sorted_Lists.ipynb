```python
class Node:
    def __init__(self, x):
        self.val = x
        self.next = None


list1 = Node(1)
list1.next = Node(2)
list1.next.next = Node(4)

list2 = Node(1)
list2.next = Node(3)
list2.next.next = Node(4)
```

## Solution 1: Two pointer

```python
class Solution(object):
    def mergeTwoLists(self, list1, list2): 
        if list1 == None:
            return list2
        if list2 == None:
            return list1

        if list1.val <= list2.val:
            head = list1
            pointer1 = list1.next
            pointer2 = list2
        else:
            head = list2
            pointer1 = list1
            pointer2 = list2.next

        current_node = head

        while pointer1 is not None and pointer2 is not None:
            if pointer1.val <= pointer2.val:
                current_node.next = pointer1
                current_node = current_node.next
                pointer1 = pointer1.next

            else:
                current_node.next = pointer2
                current_node = current_node.next
                pointer2 = pointer2.next

        if pointer1 == None:
            current_node.next = pointer2

        elif pointer2 == None:
            current_node.next = pointer1
        return head
                
sol = Solution()
print(sol.mergeTwoLists(list1, list2))
        
```

Time Complexity: $O(n+m)$ 

Space Complexity: $O(1)$

## Solution 2: Dummy node

```python
class Solution(object):
    def mergeTwoLists(self, list1, list2): 
        dummy = Node(0)
        current_node = dummy

        while list1 and list2:
            if list1.val <= list2.val:
                current_node.next = list1
                list1 = list1.next
            
            else:
                current_node.next = list2
                list2 = list2.next

            current_node = current_node.next

        if list1 == None:
            current_node.next = list2
        if list2 == None:
            current_node.next = list1
            
        return dummy.next
                
sol = Solution()
print(sol.mergeTwoLists(list1, list2))
```

Time Complexity: $O(n+m)$ 

Space Complexity: $O(1)$
